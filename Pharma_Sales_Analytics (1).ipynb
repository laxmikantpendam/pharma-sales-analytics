## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
```

## Step 2: Generate Realistic Pharma Sales Dataset
(Real pharma company data is confidential, so we build a realistic synthetic dataset with proper price ranges per drug type.)

```python
n = 2000

drug_price_map = {
    'Paracetamol': (5, 20),
    'Azithromycin': (40, 120),
    'Metformin': (10, 35),
    'Amoxicillin': (30, 90),
    'Insulin': (150, 400)
}

drug_names = list(drug_price_map.keys())
drugs = np.random.choice(drug_names, n)
unit_price = np.array([np.random.uniform(*drug_price_map[d]) for d in drugs]).round(2)

data = {
    'Order_ID': range(1001, 1001+n),
    'Date': pd.date_range('2025-01-01', periods=n, freq='6h'),
    'Drug_Name': drugs,
    'Category': np.random.choice(['Tablet', 'Syrup', 'Injection', 'Capsule'], n, p=[0.45,0.2,0.15,0.2]),
    'Region': np.random.choice(['Maharashtra','Gujarat','Delhi','Karnataka','Tamil Nadu'], n),
    'Distributor': np.random.choice(['MedPlus','Apollo Pharmacy','Wellness Forever','Netmeds'], n),
    'Quantity': np.random.randint(10, 500, n),
    'Unit_Price': unit_price,
    'Expiry_Days_Left': np.random.randint(-15, 400, n)
}

df = pd.DataFrame(data)
df['Total_Sales'] = (df['Quantity'] * df['Unit_Price']).round(2)
df['Expired_Stock'] = np.where(df['Expiry_Days_Left'] < 0, 'Yes', 'No')
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')

df.head()
```

## Step 3: Data Cleaning Checks

```python
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
print('\nData types:\n', df.dtypes)
```

## Step 4: Quick EDA — Business Insights in Python

```python
# Top selling drugs by revenue
drug_sales = df.groupby('Drug_Name')['Total_Sales'].sum().sort_values(ascending=False)
print(drug_sales)

plt.figure(figsize=(8,4))
sns.barplot(x=drug_sales.index, y=drug_sales.values, palette='viridis')
plt.title('Total Sales by Drug')
plt.ylabel('Total Sales (Rs)')
plt.xticks(rotation=30)
plt.show()
```

```python
# Expired stock — revenue at risk
expired = df[df['Expired_Stock'] == 'Yes']
print('Total expired batches:', len(expired))
print('Revenue locked in expired stock: Rs', expired['Total_Sales'].sum())

expired.groupby('Drug_Name')['Quantity'].sum().sort_values(ascending=False)
```

```python
# Region-wise sales
region_sales = df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
region_sales.plot(kind='bar', figsize=(7,4), color='teal', title='Region-wise Total Sales')
plt.ylabel('Total Sales (Rs)')
plt.show()
```

## Step 5: Export Clean Data — for SQL (SSMS) and Power BI

```python
df.to_csv('pharma_sales.csv', index=False)
print('Saved: pharma_sales.csv — import this into SSMS (SQL Server) and Power BI')
```
